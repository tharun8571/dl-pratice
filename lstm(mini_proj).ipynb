  **TOKENIZATION**

```python
token=Tokenizer()
token.fit_on_texts([data])

```

```python
total_words=__builtins__.len(token.word_index)
```

```python
total_words
```

**PREPROCESSING THE DATASET TO X AND Y, WHERE X IS ARRAY OF TOKEN AND Y IS AN ARRAY WHICH IS CATEGORIAL WITH 851(ABOVE CELL)**

```python
input_sequences = []
for sentence in data.split('\n'):
  tokenized_sentence = token.texts_to_sequences([sentence])[0]
  for i in range(1,__builtins__.len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
max_len=max([__builtins__.len(i) for i in input_sequences])
```

```python
max_len
```

PADDING THE DATASETS

```python
pad_input_sequences=pad_sequences(input_sequences,maxlen=max_len,padding='pre')
x=pad_input_sequences[:,:-1]
y=pad_input_sequences[:,-1]
```

```python
y = to_categorical(y, num_classes=total_words+1)
```

**MODEL BUILDING LSTM(LONG SHORT TERM MEMORY)**

```python
model=Sequential()
model.add(Embedding(total_words+1,100,input_length=max_len))
model.add((LSTM(150)))
model.add(Dense(total_words+1,activation='softmax'))
```

```python
model.build(input_shape=(None,max_len+1))
```

```python
model.summary()
```

**MODEL COMPILE AND FIT**

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model.fit(x,y,epochs=100)
```

**MODEL IMPORVEMENT THROUGH EXTRA EPOCHS**

```python
model.fit(x,y,epochs=200,initial_epoch=99)
```

**PREDICTION**

```python
text="""The internship program """
for i in range(10):
  tokenized_text=token.texts_to_sequences([text])
  pad_text=pad_sequences(tokenized_text,maxlen=max_len,padding='pre')
  pos=np.argmax(model.predict(pad_text))
  for word,index in token.word_index.items():
    if index==pos:
      text=text+" "+word
      print(text)
```
